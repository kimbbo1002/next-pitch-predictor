## Sample week findings (2025-06-01 to 2025-06-07)

### 1. Size
- **27,121 rows x 119 columns**, 30.0 MMB in memory (4.6 MB as Parquet on disk)
- Each row is **one pitch**.

### 2. Games
- **94 games**, averaging **288 pitches per game** (min 206, max 354, std 31).

### 3. Game Types
- All rows are `R` (regular season).
- Full season pull should still filter on `game_type == "R"`, in case spring training or postseason games appear.

### 4. **Pitch Types**
- **16 distinct types + 83 missing (0.3%).**
- Common (>2%): FF 30.9%, SI 16.1%, SL 14.8%, CH 10.5%, FC 7.7%, ST 6.9%, CU 6.6%, FS 2.9%, KC 2.1%.
- Rare (<1%): SV 0.6%, FO 0.18%, FA 0.12%, EP 0.10%, CS 0.08%, KN 0.01%, PO 0.01%.
- 'PO' is a **pitchout**, a deliberate throw wide, not a real pitch type.
- **Next Step:** decide label mapping (merge similar types), drop position player pitchers and pitchouts, handle missing labels.

### 5. Unique Key
- `(game_pk, at_bat_number, pitch_number)` has **0 duplicates**, therefore uniquely identifies a pitch and can be use for sorting and validation.

### 6. Row order
- Rows are in **reverse chronological order**.
- **Leakage risk:** history features computed without sorting first would actually use the *next* pitch. Always sort by `game_date, game_pk, at_bat_number, pitch_numer` ascending before building lag features.

### 7. Full-season estimate
- 2,430 games x 288 pitches ≈ **700,000 pitches**, about 25.8x this sample.
- Memory: 30 MB × 25.8 ≈ **775 MB** with all 119 columns, which is manageable but heavy. Keeping only the needed columns (about 30) should cut this.
- Disk: ~120 MB as Parquet.
